```python
# ============================================================
# REPRESENTATIVE LAP AUDIT
# ============================================================

import numpy as np
import pandas as pd

audit = main.copy()

# ------------------------------------------------------------
# 1. Create within-stint statistics
# ------------------------------------------------------------

stint_cols = [
    "Year",
    "Event",
    "SessionName",
    "Driver",
    "StintID"
]

audit["StintLapMedian"] = (
    audit.groupby(stint_cols)["LapTimeSeconds"]
    .transform("median")
)

audit["StintLapMAD"] = (
    audit.groupby(stint_cols)["LapTimeSeconds"]
    .transform(
        lambda x: np.median(np.abs(x - np.median(x)))
    )
)

# Avoid division by zero
audit["StintLapMAD"] = audit["StintLapMAD"].replace(0, np.nan)

# Robust deviation from stint median
audit["LapRobustDeviation"] = (
    np.abs(audit["LapTimeSeconds"] - audit["StintLapMedian"])
    / audit["StintLapMAD"]
)

# ------------------------------------------------------------
# 2. Telemetry deviation
# ------------------------------------------------------------

audit["StintSpeedMedian"] = (
    audit.groupby(stint_cols)["TelemetryMeanSpeed_V4"]
    .transform("median")
)

audit["SpeedDeviationPct"] = (
    (audit["TelemetryMeanSpeed_V4"] - audit["StintSpeedMedian"])
    / audit["StintSpeedMedian"]
)

# ------------------------------------------------------------
# 3. Basic telemetry quality
# ------------------------------------------------------------

audit["TelemetryQuality"] = (
    audit["TelemetrySampleCount_V4"].notna()
    &
    audit["TelemetryMeanSpeed_V4"].notna()
    &
    audit["TelemetryMeanThrottle_V4"].notna()
)

# ------------------------------------------------------------
# 4. Candidate suspicious laps
# ------------------------------------------------------------

audit["SuspiciousLap"] = (
    (audit["LapRobustDeviation"] > 5)
    |
    (
        (audit["SpeedDeviationPct"] < -0.20)
        &
        (audit["LapRobustDeviation"] > 2)
    )
)

# ------------------------------------------------------------
# 5. Summary
# ------------------------------------------------------------

print("===== REPRESENTATIVE LAP AUDIT =====")

print("\nTotal rows:")
print(len(audit))

print("\nTelemetry quality:")
print(audit["TelemetryQuality"].value_counts())

print("\nSuspicious laps:")
print(audit["SuspiciousLap"].value_counts())

print("\nSuspicious percentage:")
print(
    round(
        audit["SuspiciousLap"].mean() * 100,
        2
    ),
    "%"
)

# ------------------------------------------------------------
# 6. Show worst candidates
# ------------------------------------------------------------

worst = (
    audit[audit["SuspiciousLap"]]
    .sort_values(
        ["LapRobustDeviation"],
        ascending=False
    )
)

print("\n===== WORST SUSPICIOUS LAPS =====")

print(
    worst[
        [
            "Year",
            "Event",
            "SessionName",
            "Driver",
            "StintID",
            "LapNumber",
            "TyreAge",
            "Compound",
            "LapTimeSeconds",
            "StintLapMedian",
            "LapRobustDeviation",
            "TelemetryMeanSpeed_V4",
            "SpeedDeviationPct",
            "TelemetryMeanThrottle_V4",
            "TelemetryFullThrottleFraction_V4",
            "TrafficExposureLevel_V4"
        ]
    ].head(30).to_string(index=False)
)
```

```python
# ============================================================
# SUSPICIOUS LAPS BY TYRE AGE
# ============================================================

print("===== SUSPICIOUS LAPS BY TYRE AGE =====")

age_summary = (
    audit
    .groupby("TyreAge")
    .agg(
        TotalLaps=("LapTimeSeconds", "size"),
        SuspiciousLaps=("SuspiciousLap", "sum"),
        MedianLapTime=("LapTimeSeconds", "median"),
        MedianSpeed=("TelemetryMeanSpeed_V4", "median"),
        MedianThrottle=("TelemetryMeanThrottle_V4", "median")
    )
)

age_summary["SuspiciousPct"] = (
    age_summary["SuspiciousLaps"]
    / age_summary["TotalLaps"]
    * 100
)

print(age_summary.round(2).to_string())


# ============================================================
# SUSPICIOUS LAPS BY COMPOUND
# ============================================================

print("\n===== SUSPICIOUS LAPS BY COMPOUND =====")

compound_summary = (
    audit
    .groupby("Compound")
    .agg(
        TotalLaps=("LapTimeSeconds", "size"),
        SuspiciousLaps=("SuspiciousLap", "sum"),
        MedianLapTime=("LapTimeSeconds", "median"),
        MedianSpeed=("TelemetryMeanSpeed_V4", "median")
    )
)

compound_summary["SuspiciousPct"] = (
    compound_summary["SuspiciousLaps"]
    / compound_summary["TotalLaps"]
    * 100
)

print(compound_summary.round(2).to_string())


# ============================================================
# SUSPICIOUS LAPS BY SESSION
# ============================================================

print("\n===== SUSPICIOUS LAPS BY SESSION =====")

session_summary = (
    audit
    .groupby("SessionName")
    .agg(
        TotalLaps=("LapTimeSeconds", "size"),
        SuspiciousLaps=("SuspiciousLap", "sum")
    )
)

session_summary["SuspiciousPct"] = (
    session_summary["SuspiciousLaps"]
    / session_summary["TotalLaps"]
    * 100
)

print(session_summary.round(2).to_string())
```

```python
# ============================================================
# INVESTIGATE THE TYRE AGE 3 ANOMALY
# ============================================================

# Find stints where we have consecutive tyre ages 2, 3 and 4
stint_cols = [
    "Year",
    "Event",
    "SessionName",
    "Driver",
    "StintID"
]

age_sequence = (
    main.groupby(stint_cols)["TyreAge"]
    .agg(lambda x: set(x.dropna()))
)

candidate_stints = age_sequence[
    age_sequence.apply(lambda x: {2.0, 3.0, 4.0}.issubset(x))
].index

print("Candidate stints:", len(candidate_stints))


# ------------------------------------------------------------
# Pick 20 candidate stints randomly
# ------------------------------------------------------------

rng = np.random.default_rng(42)

n = min(20, len(candidate_stints))

selected = candidate_stints[
    rng.choice(len(candidate_stints), size=n, replace=False)
]


# ------------------------------------------------------------
# Print consecutive laps around age 2-4
# ------------------------------------------------------------

for stint in selected:

    mask = np.ones(len(main), dtype=bool)

    for col, value in zip(stint_cols, stint):
        mask &= main[col].eq(value)

    example = (
        main.loc[mask]
        .sort_values("LapNumber")
        [
            [
                "Year",
                "Event",
                "SessionName",
                "Driver",
                "StintID",
                "LapNumber",
                "TyreAge",
                "Compound",
                "LapTimeSeconds",
                "TelemetryMeanSpeed_V4",
                "TelemetryMeanThrottle_V4",
                "TelemetryFullThrottleFraction_V4",
                "TelemetryMeanBrake_V4",
                "TrafficExposureLevel_V4",
                "TrafficExposureScore_V4"
            ]
        ]
    )

    # Only show ages 1-6 so the transition is easy to inspect
    example = example[
        example["TyreAge"].between(1, 6)
    ]

    print("\n" + "=" * 100)
    print("STINT:", stint)
    print("=" * 100)

    print(example.to_string(index=False))
```

```python
# ============================================================
# CLEAN REPRESENTATIVE LAP DATASET — VERSION 1
# ============================================================

import numpy as np
import pandas as pd

clean = main.copy()

stint_cols = [
    "Year",
    "Event",
    "SessionName",
    "Driver",
    "StintID"
]

# ------------------------------------------------------------
# 1. Basic telemetry validity
# ------------------------------------------------------------

clean["TelemetryValid"] = (
    clean["TelemetryMeanSpeed_V4"].notna()
    &
    clean["TelemetryMeanThrottle_V4"].notna()
    &
    clean["TelemetryFullThrottleFraction_V4"].notna()
    &
    clean["TelemetrySampleCount_V4"].notna()
)

# ------------------------------------------------------------
# 2. Calculate robust stint baselines
# ------------------------------------------------------------

for col in [
    "LapTimeSeconds",
    "TelemetryMeanSpeed_V4",
    "TelemetryMeanThrottle_V4",
    "TelemetryFullThrottleFraction_V4"
]:

    clean[f"{col}_StintMedian"] = (
        clean.groupby(stint_cols)[col]
        .transform("median")
    )

# ------------------------------------------------------------
# 3. Relative telemetry / pace measures
# ------------------------------------------------------------

clean["LapTimeRatio"] = (
    clean["LapTimeSeconds"]
    / clean["LapTimeSeconds_StintMedian"]
)

clean["SpeedRatio"] = (
    clean["TelemetryMeanSpeed_V4"]
    / clean["TelemetryMeanSpeed_V4_StintMedian"]
)

clean["ThrottleRatio"] = (
    clean["TelemetryMeanThrottle_V4"]
    / clean["TelemetryMeanThrottle_V4_StintMedian"]
)

clean["FullThrottleRatio"] = (
    clean["TelemetryFullThrottleFraction_V4"]
    / clean["TelemetryFullThrottleFraction_V4_StintMedian"]
)

# ------------------------------------------------------------
# 4. Identify clearly non-representative laps
#
# We deliberately use multiple signals.
#
# Slow lap + slow telemetry = strong evidence.
# ------------------------------------------------------------

clean["NonRepresentative"] = False

# Strong slow-lap condition
strong_slow = (
    (clean["LapTimeRatio"] > 1.20)
    &
    (clean["SpeedRatio"] < 0.85)
    &
    (clean["ThrottleRatio"] < 0.70)
)

# Extremely weak driving intensity
very_low_intensity = (
    (clean["SpeedRatio"] < 0.75)
    &
    (clean["ThrottleRatio"] < 0.60)
    &
    (clean["FullThrottleRatio"] < 0.50)
)

clean.loc[
    strong_slow | very_low_intensity,
    "NonRepresentative"
] = True

# ------------------------------------------------------------
# 5. Remove laps with no usable telemetry
# ------------------------------------------------------------

clean.loc[
    ~clean["TelemetryValid"],
    "NonRepresentative"
] = True

# ------------------------------------------------------------
# 6. Create final clean dataset
# ------------------------------------------------------------

clean_v1 = clean[
    ~clean["NonRepresentative"]
].copy()

# ------------------------------------------------------------
# 7. Summary
# ------------------------------------------------------------

print("===== CLEAN DATASET V1 =====")

print("Original rows:", len(clean))

print("Clean rows:", len(clean_v1))

print(
    "Removed:",
    len(clean) - len(clean_v1)
)

print(
    "Removed percentage:",
    round(
        (1 - len(clean_v1) / len(clean)) * 100,
        2
    ),
    "%"
)

print("\nRemoved by reason:")
print(
    clean["NonRepresentative"]
    .value_counts()
)

# ------------------------------------------------------------
# 8. Tyre-age distribution after cleaning
# ------------------------------------------------------------

print("\n===== CLEAN LAPS BY TYRE AGE =====")

age_clean = (
    clean_v1
    .groupby("TyreAge")
    .size()
    .rename("CleanLaps")
)

print(age_clean.to_string())

# ------------------------------------------------------------
# 9. Compound distribution
# ------------------------------------------------------------

print("\n===== CLEAN LAPS BY COMPOUND =====")

print(
    clean_v1["Compound"]
    .value_counts()
)

# ------------------------------------------------------------
# 10. Check age-3 specifically
# ------------------------------------------------------------

print("\n===== AGE 3 AFTER CLEANING =====")

age3_clean = clean_v1[
    clean_v1["TyreAge"] == 3
]

print("Age 3 clean laps:", len(age3_clean))

print(
    age3_clean[
        [
            "LapTimeSeconds",
            "TelemetryMeanSpeed_V4",
            "TelemetryMeanThrottle_V4",
            "TelemetryFullThrottleFraction_V4"
        ]
    ]
    .describe()
    .round(2)
)
```
